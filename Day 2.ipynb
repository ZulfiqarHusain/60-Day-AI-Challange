```python
# Print a cleaned, lowercase version of an input text (strip punctuation, normalize whitespace)

import re
def clean_text(input_string):
  #Only Lowercase for uniformity
  text = input_string.lower()
  # Use Regex for removing Punctuation
  text = re.sub(r'[^\w\s]', '', text)
  #Remove unnecessary White Spaces
  cleaned = " ".join(text.split())
  return cleaned

sentence = "Bhai, AI is Amazing! Let's Start Learning... 100%"
print(f"original:{sentence}")
print(f"cleaned:{clean_text(sentence)}")
```

```python
#Write a script that reads a sentence and counts word frequency

def count_word_frequency(sentence):
  #First Clean text of cell 1
  clean_s = clean_text(sentence)
  words = clean_s.split()

  frequency = {}
  for word in words:
    frequency[word] = frequency.get(word, 0) + 1
  return frequency

test_sentence ="ai is great and its our future"
print("Word Frequency Result:")
print(count_word_frequency(test_sentence))

```
