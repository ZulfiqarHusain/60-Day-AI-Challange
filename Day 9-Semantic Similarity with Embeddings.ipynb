```python
from sentence_transformers import SentenceTransformer, util

# 1. Pre-trained Model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Test Sentences
sentences = [
    "The cat sits outside",
    "A man is playing guitar",
    "The feline is resting outdoors",
    "He is strumming a musical instrument"
]

# 3. Sentences into Vectors (Numbers)
embeddings = model.encode(sentences)

# 4. Similarity Compute
# First sentence (Cat) vs Third sentence (Feline)
sim1 = util.cos_sim(embeddings[0], embeddings[2])
# Second sentence (Guitar) vs Fourth sentence (Instrument)
sim2 = util.cos_sim(embeddings[1], embeddings[3])

print(f"Similarity (Cat vs Feline): {sim1.item():.4f}")
print(f"Similarity (Guitar vs Instrument): {sim2.item():.4f}")
```
