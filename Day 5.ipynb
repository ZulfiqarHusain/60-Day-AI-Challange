```python
from sentence_transformers import SentenceTransformer, util
import torch

# 1. Model Load
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Dataset
sentences = [
    "Artificial Intelligence is transforming the world.",
    "Machine learning models require a lot of data.",
    "Python is the most popular language for AI.",
    "Deep learning is a subset of machine learning.",
    "Natural Language Processing helps computers understand humans.",
    "Computer vision is used in self-driving cars.",
    "Neural networks are inspired by the human brain.",
    "Generative AI can create art and text.",
    "Large Language Models like GPT are very powerful.",
    "Reinforcement learning is used in robotics.",
    "Data science involves extracting insights from data.",
    "Cloud computing provides scalable resources for AI.",
    "Edge AI runs models on local devices.",
    "Ethics in AI is a crucial topic for researchers.",
    "Supervised learning uses labeled datasets.",
    "Unsupervised learning finds hidden patterns.",
    "Transformers are the backbone of modern NLP.",
    "Quantum computing could speed up AI training.",
    "AI chatbots are improving customer service.",
    "Automation reduces manual effort in industries."
]

# 3. Embedding: Sentences Change Into Numbers
# AI Understand Meaning Of The Whole Words
sentence_embeddings = model.encode(sentences, convert_to_tensor=True)

# 4. Search Function
def semantic_search(query):
    # Query Into Number
    query_embedding = model.encode(query, convert_to_tensor=True)

    # Check Cosine Similarity
    cosine_scores = util.cos_sim(query_embedding, sentence_embeddings)[0]

    # Find Top 3 results
    top_results = torch.topk(cosine_scores, k=3)

    print(f"\nQuery: {query}")
    print("-" * 30)
    for score, idx in zip(top_results[0], top_results[1]):
        print(f"Score: {score:.4f} | {sentences[idx]}")

# 5. Testing the Engine
my_query = "How do computers talk to people?"

semantic_search(my_query)
```
