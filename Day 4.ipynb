```python
!pip install sentence-transformers
from sentence_transformers import SentenceTransformer, util
import torch

model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
sentences = [
    "The cat sits outside",
    "A man is playing guitar",
    "The feline is resting outdoors",  # it's means like a cat sentence
    "The new movie is awesome",
    "A person is strumming a musical instrument" # it's similar like a guitar
]

# Convert Sentences into a number
embeddings = model.encode(sentences)

print("Embedding generate successfully!")
```

```python
import torch

# Comparison pairs
# Pair 1: Cat vs Feline (Should be high)
# Pair 2: Cat vs Movie (Should be low)

score1 = util.cos_sim(embeddings[0], embeddings[2])
score2 = util.cos_sim(embeddings[0], embeddings[3])

print(f"Similarity (Cat vs Feline): {score1.item():.4f}")
print(f"Similarity (Cat vs Movie): {score2.item():.4f}")
```
