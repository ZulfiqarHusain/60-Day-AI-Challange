```python
import pandas as pd
import re
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# 1. Essential Resources Download
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
ps = PorterStemmer()

# 2. Preprocessing Function
def clean_text(text):
    # Lowercase
    text = text.lower()
    # Remove Special Characters/Numbers
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    # Tokenize, Remove Stopwords & Stemming
    words = text.split()
    cleaned = [ps.stem(w) for w in words if w not in stop_words]
    return " ".join(cleaned)

# 3. Simple Dataset
data = {
    'text': [
        'I loved the movie, it was amazing!',
        'The acting was terrible and waste of time.',
        'A masterpiece of cinema, truly great plot.',
        'Worst experience ever, very boring.',
        'I really enjoyed the performance.',
        'Not good, very slow and frustrating.'
    ],
    'sentiment': ['pos', 'neg', 'pos', 'neg', 'pos', 'neg']
}

df = pd.DataFrame(data)

# 4. Cleaning Apply
df['cleaned_text'] = df['text'].apply(clean_text)
print("--- Preprocessed Data ---")
print(df[['text', 'cleaned_text']])

# 5. Vectorization & Training
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['cleaned_text'])
y = df['sentiment']

model = MultinomialNB()
model.fit(X, y)

# 6. Real-time Testing
test_sentences = [
    "I am loving this so much",
    "It was not a good movie",
    "Absolutely waste of money",
    "The plot was incredibly amazing"
]

print("\n--- Predictions ---")
for s in test_sentences:
    cleaned_s = clean_text(s)
    vec_s = vectorizer.transform([cleaned_s])
    prediction = model.predict(vec_s)[0]
    print(f"Original: {s} | Predicted: {prediction}")
```
