```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Simple Dataset
data = {
    'text': [
        'I love this movie', 'Great plot and acting', 'Amazing experience',
        'Worst movie ever', 'Waste of time', 'I hated the acting',
        'Very boring and slow', 'A masterpiece', 'Not good at all'
    ],
    'sentiment': ['positive', 'positive', 'positive', 'negative', 'negative', 'negative', 'negative', 'positive', 'negative']
}

df = pd.DataFrame(data)

# 2. Text to Numbers (Vectorization)
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(df['text'])
y = df['sentiment']

# 3. Model Training
model = MultinomialNB()
model.fit(X, y)

# 4. Manual Testing
test_samples = [
    "The movie was great",
    "I did not like it",
    "Truly amazing",
    "It was a boring film",
    "Excellent work"
]

test_vectorized = vectorizer.transform(test_samples)
predictions = model.predict(test_vectorized)

# Results
for text, pred in zip(test_samples, predictions):
    print(f"Text: {text} | Sentiment: {pred}")
```
