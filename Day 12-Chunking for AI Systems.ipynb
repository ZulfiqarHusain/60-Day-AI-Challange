```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

# 1. Hamara Sample Long Text (About your AI journey)
text = """Zulfiqar Husain is an AI & DS student at SIRT-E Bhopal.
Currently, he is documenting his 60-Day AI Challenge.
Day 11 was about Retrieval Systems where he used TF-IDF and Cosine Similarity.
Day 12 is about Chunking, which is a crucial step in building efficient RAG pipelines.
Chunking helps LLMs handle large contexts by breaking them into manageable pieces
without losing the core meaning of the information."""

# 2. Setup Recursive Chunking
# chunk_size
# chunk_overlap
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 100,
    chunk_overlap = 20
)

# 3. Create Chunks
chunks = text_splitter.split_text(text)

# 4. Results Output
print(f"Total Chunks Created: {len(chunks)}\n")
for i, chunk in enumerate(chunks):
    print(f"Chunk {i+1}:\n{chunk}\n{'-'*20}")
```
