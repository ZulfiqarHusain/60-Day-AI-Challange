```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report

# 1. Dataset )
data = {
    'text': [
        'I love this product, it is amazing', 'Great service and friendly staff',
        'Worst experience ever, hated it', 'The quality is very poor',
        'This is the best thing I have ever bought',
        'Excellent work, highly recommend', 'I am disappointed with the results',
        'Simply brilliant and worth every penny', 'Avoid this, it is a complete waste'
    ],
    'label': ['positive', 'positive', 'negative', 'negative', 'positive', 'positive', 'negative', 'positive', 'negative']
}
df = pd.DataFrame(data)

# 2. Creating the Pipeline
# We can use TfidfVectorizer which is better then simple count
nlp_pipeline = Pipeline([
    ('tfidf', TfidfVectorizer(stop_words='english')),
    ('clf', MultinomialNB())
])

# 3. Train the Pipeline
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42)
nlp_pipeline.fit(X_train, y_train)

# 4. Testing the whole Pipeline with new input
new_comments = [
    "This is the best thing I have ever bought",
    "Worst experience ever, hated "
]

predictions = nlp_pipeline.predict(new_comments)

for text, pred in zip(new_comments, predictions):
    print(f"Comment: {text} | Result: {pred}")
```
