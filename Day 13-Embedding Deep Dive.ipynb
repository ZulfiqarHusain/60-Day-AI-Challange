```python
from sentence_transformers import SentenceTransformer
import numpy as np

# 1. Load a Pre-trained Model
model = SentenceTransformer('all-MiniLM-L6-v2')

# 2. Sentences to Compare
sentences = [
    "I'm Self Obsessed Person.",
    "I love cars.",
    "Bhopal is a beautiful city with many lakes.",
    "The weather is very pleasant today."
]

# 3. Generate Embeddings
embeddings = model.encode(sentences)

# 4. Deep Dive Observations
for i, sentence in enumerate(sentences):
    print(f"Sentence: {sentence}")
    print(f"Vector Shape: {embeddings[i].shape}") # it's finds dimension
    print(f"First 5 Values: {embeddings[i][:5]}") # Glimpse a vector
    print("-" * 30)

# 5. Semantic Check (Distance calculation)
from sklearn.metrics.pairwise import cosine_similarity
sim = cosine_similarity([embeddings[0]], [embeddings[1]])
print(f"Similarity between AI sentences: {sim[0][0]:.4f}")
```
