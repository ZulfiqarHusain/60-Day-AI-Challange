```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

nltk.download('punkt')
nltk.download('stopwords')
nltk.download('punkt_tab')
```

```python
def preprocess_text(text):
    print(f"--- Raw Input ---\n{text}")

    # Step 1: Tokenization (Breaking into words)
    tokens = word_tokenize(text.lower())
    print(f"\n1. Tokens:\n{tokens}")

    # Step 2: Remove Stop words and Punctuation
    stop_words = set(stopwords.words('english'))
    punctuation = set(string.punctuation)

    cleaned_tokens = [w for w in tokens if w not in stop_words and w not in punctuation]
    print(f"\n2. Cleaned (No Stopwords/Punctuation):\n{cleaned_tokens}")

    return " ".join(cleaned_tokens)

# Sample text
sample_paragraph = "Machines don't read words. They read numbers."
cleaned_sample = preprocess_text(sample_paragraph)
```

```python
# Sample dataset for the Bag-of-Words model
corpus = [
    "Machines read numbers not words.",
    "Preprocessing turns words into numbers.",
    "Every NLP system starts with preprocessing."
]

# Clean the corpus first
cleaned_corpus = [" ".join([w for w in word_tokenize(doc.lower())
                            if w not in set(stopwords.words('english'))
                            and w not in set(string.punctuation)])
                  for doc in corpus]

# Convert to Bag-of-Words
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(cleaned_corpus)

# Display as a DataFrame for clear submission
feature_names = vectorizer.get_feature_names_out()
df_bow = pd.DataFrame(bow_matrix.toarray(), columns=feature_names)

print("\n--- Final Preprocessed Dataset (Bag-of-Words) ---")
display(df_bow)
```
